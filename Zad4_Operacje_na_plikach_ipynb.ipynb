Wersja po lini

```python
import os

filesource="numbers.txt"
filedestiny="copy.txt"

with open(filesource,"r") as f:
  with open(filedestiny,"w") as f1:
    for line in f:
      f1.write(line)
```

```python
with open(filesource, "r") as f:
  content = f.read()
  with open("copy1.txt","w") as f1:
    f1.write(content)
```
