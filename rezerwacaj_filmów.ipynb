```python

movies = {
    "Finding Nemo": [5, 2],
    "Moana": [6, 3],
    "Batman": [18, 5],
    "The Lion King": [10, 4]
}

print("Witaj w systemie rezerwacji kinowej!")

while True:
    print("\n--- DOSTĘPNE FILMY ---")


    for tytul, dane in movies.items():
        miejsca = dane[0]
        wiek = dane[1]
        print(f"Film: {tytul} | Wolne miejsca: {miejsca} | Wiek: od {wiek} lat")

    print("----------------------")
    choice = input("Wpisz tytuł filmu (lub 'exit' aby wyjść): ")

    # WYJŚCIE
    if choice == 'exit':
        print("Dziękujemy! Do widzenia.")
        break


    if choice in movies:

        movie_data = movies[choice]
        available_seats = movie_data[0]
        min_age = movie_data[1]

        #Sprawdzenie wieku
        try:
            user_age = int(input("Ile masz lat? "))

            if user_age < min_age:
                print(f"Przykro mi, ten film jest tylko dla widzów od {min_age} lat.")
                continue

            # Pytamy o bilety
            tickets = int(input(f"Ile biletów na '{choice}' chcesz kupić? "))

            if tickets <= available_seats:
                new_seats_count = available_seats - tickets
                movies[choice] = [new_seats_count, min_age]
                print(f"Sukces! Kupiono {tickets} biletów na {choice}.")
            else:
                print(f"Niestety, mamy tylko {available_seats} wolnych miejsc.")

        except ValueError:
            print("Błąd: Proszę wpisywać tylko liczby!")

    else:
        print("Nie ma takiego filmu w repertuarze.")
```
