```python
import os
filename ="plik.txt"

if os.path.exists(filename):
  print("Plik istnieje!")
else:
  with open(filename,"w") as f:
    f.write("to jest plik tekstowy ")
  print("Plik utworzony\n Dane zapisane!")
```

```python
with open("plik.txt", "r") as f:
  content = f.read()
  print(content)
```
