```python
import os

filename="raw.txt"
filename1="cleare.txt"

with open(filename, 'r') as f:
  with open(filename1, 'w') as f1:
    for line in f:
      if line.strip()!="":
        f1.write(line)

with open(filename1, 'r') as f1o:
  print("-----------\n")
  print("Zawartosc plikui po czyszczeniu:\n")
  print(f1o.read())
```
